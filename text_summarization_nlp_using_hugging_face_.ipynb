```python
import datasets
import transformers
from transformers import pipeline
```

```python
from datasets import load_dataset
dataset = load_dataset("abisee/cnn_dailymail", "3.0.0")
```

```python
dataset
```

```python
 pipe =pipeline("summarization",model = "facebook/bart-large-cnn")
```

```python
input_text = dataset['train'][0]["article"][:2000]
```

```python
print(input_text)
```

```python
pipe_out = pipe(input_text , max_length= 200, clean_up_tokenization_spaces = True)
```
